# 04 — Validation against published results

Three per-rule validations (paper Table 1, `validation/`):

1. **Smith (1993)** — the comparator must reproduce the closed-form
   wake-ingestion limits;
2. **MIT/Aurora D8** (Uranga et al. 2017) — measured 8.2% ± 0.8% power saving
   with the equal-nozzle-area rule, and the 5.2/2.4/0.6 mechanism split;
3. **NASA STARC-ABL** (Welstead & Felder 2016; Yildirim et al. 2022) — the
   net benefit must fall inside the published CFD bracket.

Nothing here is tuned to match: where the low-order model misses, the paper
diagnoses rather than fudges.
