# 05 — Reproduce every paper figure

One cell per figure. All eight figures of the manuscript regenerate from the
cached Parquet data in `data/` (UQ figures automatically prefer the
production-density run under `data/production/` when present, falling back to
the committed pilot data). Outputs land in `figures/` as PDF + PNG — exactly
the files `paper/main.tex` includes.


```python
# resolve the repo root so the notebook runs from notebooks/ or the repo root
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
while not (ROOT / "pyproject.toml").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "studies"))
import plotstyle

plotstyle.apply()
import figures
from IPython.display import Image

FIG = plotstyle.FIGDIR
# use production UQ data when the (git-ignored) production run exists
tag = "_prod" if (ROOT / "data" / "production" /
                  "uq_prod_sobol_psc_net.parquet").exists() else ""
print(f"UQ tag: {tag!r} ({'production N=2^14' if tag else 'pilot N=1024'})")
```
